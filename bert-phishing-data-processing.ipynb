### Remove empty columns

```python
import pandas as pd

# Step 1: Load the CSV file
df = pd.read_csv('CEAS_08.csv')

# Step 2: Identify the index of the "label" column
label_idx = df.columns.get_loc('label')

# Step 3: Remove empty columns after the "label" column
df = df.iloc[:, :label_idx+1].copy()

# Step 4: Save the cleaned DataFrame to a new CSV file
df.to_csv('CEAS_08.csv', index=False)
```

### Change labels from float to integer

```python
import pandas as pd

# Step 1: Load the CSV file
df = pd.read_csv('CEAS_08_5.csv')

# Step 2: Convert the 'label' column from float to integer
# Replace 'label' with the actual column name if it's different
df['label'] = df['label'].astype(float).astype(int)

# Step 3: Save the updated DataFrame to a new CSV file
df.to_csv('CEAS_08_6.csv', index=False)
```

### Change the encodings of the file

```python
# Specify the path to the input file and output file
input_file_path = 'datasets I have worked on\\spam.csv'
output_file_path = 'datasets I have worked on\\spam_utf8.csv'

# Original encoding
original_encoding = 'Windows-1252'

# New encoding
new_encoding = 'utf-8'

# Step 1: Read the file with the original encoding
with open(input_file_path, 'r', encoding=original_encoding) as file:
    content = file.read()

# Step 2: Write the content to a new file with the new encoding
with open(output_file_path, 'w', encoding=new_encoding) as file:
    file.write(content)

print(f"File has been converted from {original_encoding} to {new_encoding} and saved as {output_file_path}.")
```

### Remove empty space in the text itself

```python
import pandas as pd

# Step 1: Load the CSV file
df = pd.read_csv('CEAS_08_3.csv')

# Step 2: Clean the text in the specified column (e.g., 'body')
df['body'] = df['body'].str.replace('\s+', ' ', regex=True).str.strip()

# Step 3: Save the cleaned DataFrame to a new CSV file
df.to_csv('CEAS_08_4.csv', index=False)
```

### Remove any row that didn't have a text and the label

```python
import pandas as pd

# Step 1: Load the CSV file
df = pd.read_csv('CEAS_08_4.csv')

# Step 2: Filter rows
# Keep rows where the first and second columns are not empty (non-NaN and non-empty strings)
# and the third column contains either 1 or 0

filtered_df = df[
    (df.iloc[:, 0].notna()) & (df.iloc[:, 0] != '') &  # First column must have text
    (df.iloc[:, 1].isin(['0.0', '1.0']))                       # Third column must be 1 or 0
]

# Step 3: Save the filtered DataFrame to a new CSV file
filtered_df.to_csv('CEAS_08_5.csv', index=False)

```

### Removes URLs and special characters

```python
import pandas as pd
import re

# Step 1: Load the CSV file
df = pd.read_csv('CEAS_08_1.csv')

# Step 2: Remove special characters from the 'body' column
df['subject'] = df['subject'].apply(lambda x: re.sub(r'[^A-Za-z0-9\s]', '', str(x)))
df['body'] = df['body'].apply(lambda x: re.sub(r'http\S+|www\S+|https\S+', '', str(x), flags=re.MULTILINE))
df['body'] = df['body'].apply(lambda x: re.sub(r'[^A-Za-z0-9\s]', '', str(x)))

# Step 3: Save the cleaned DataFrame to a new CSV file
df.to_csv('CEAS_08_2.csv', index=False)
```

### Merges the text of two columns into one

```python
import pandas as pd

# Step 1: Load the CSV file
df = pd.read_csv('CEAS_08_2.csv')

# Step 2: Merge the 'subject' and 'body' columns
df['body'] = df['subject'].fillna('') + ', ' + df['body'].fillna('')

# Step 3: (Optional) Drop the 'subject' column if no longer needed
df = df.drop(columns=['subject'])

# Step 4: Save the merged DataFrame to a new CSV file
df.to_csv('CEAS_08_3.csv', index=False)
```

### Switches format from text, label to label, text

```python
import pandas as pd

# Step 1: Load the CSV file
df = pd.read_csv('CEAS_08_6.csv')

# Step 2: Switch two columns
# Replace 'column1' and 'column2' with the actual column names
df = df[['label', 'text'] + [col for col in df.columns if col not in ['text', 'label']]]

# Step 3: Save the updated DataFrame to a new CSV file
df.to_csv('CEAS_08_7.csv', index=False)
```

### Orders the instances in ascending order according to label

```python
import pandas as pd

# Step 1: Load the CSV file
df = pd.read_csv('CEAS_08_7.csv')

# Step 2: Sort the rows in ascending order based on the 'label' column
# Replace 'label' with the actual column name you want to sort by
df_sorted = df.sort_values(by='label', ascending=True)

# Step 3: Save the sorted DataFrame to a new CSV file
df_sorted.to_csv('CEAS_08_7.csv', index=False)
```
